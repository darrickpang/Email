```python
from ultralytics import YOLO
```

```python
model = YOLO("yolov8s.pt")  # or yolov8n.pt for extra speed
```

```python
from google.colab import files
uploaded = files.upload()
```

```python
filename = list(uploaded.keys())[0]   # first file uploaded
print("File name:", filename)
```

```python
results = model.predict(list(uploaded.keys())[0], show=True)
```
