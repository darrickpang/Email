```python
from ultralytics import YOLO
import cv2
import matplotlib.pyplot as plt
```

```python
model = YOLO("yolov8s.pt")  # or yolov8n.pt for extra speed
```

```python
img = cv2.imread("highway.jpg")            # replace with dataset image
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
plt.imshow(cv2.cvtColor(gray, cv2.COLOR_BGR2RGB))
plt.title("Original Road Image")
plt.show()
```

```python
results = model.predict("highway.jpg", show=True)
```

```python
results = model.predict(
    source="video.mp4",
    save=True,       # saves the output video
    show=False       # Colab cannot display video frames in real-time
)
```

```python
for frame in results:
    print(frame.boxes)  # bounding boxes and detections
```
