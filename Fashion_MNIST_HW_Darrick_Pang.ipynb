````python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, models
import matplotlib.pyplot as plt
from tensorflow.keras import optimizers
from tensorflow.keras import regularizers
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
import pandas as pd
from tensorflow.keras.constraints import max_norm
import pandas as pd
````

````python
fashion_mnist = keras.datasets.fashion_mnist

(X_train_full, y_train_full), (X_test, y_test) = fashion_mnist.load_data()
X_valid, X_train = X_train_full[:5000] / 255., X_train_full[5000:] / 255.
y_valid, y_train = y_train_full[:5000], y_train_full[5000:]
X_test = X_test / 255.

results = []
````

````python
results = []
weight_decay = 1e-4
lr = "N/A"

def save_model_results(model, name, lr):
  model.compile(loss="sparse_categorical_crossentropy", optimizer="sgd", metrics=["accuracy"])

  history = model.fit(X_train, y_train, epochs=30, validation_data=(X_valid, y_valid))

  test_loss, test_acc = model.evaluate(X_test, y_test)

  results.append({"Name": name, "Accuracy": float(test_acc), "Loss": float(test_loss), "Learning rate": lr})

  pd.DataFrame(history.history).plot(figsize=(8, 5))
  plt.grid(True)
  plt.gca().set_ylim(0, 1)
  # save_fig("keras_learning_curves_plot")
  plt.show()

  return {"Name": name, "Accuracy": float(test_acc), "Loss": float(test_loss), "Learning rate": lr}

model_original = models.Sequential([
    layers.Input(shape=(28, 28)),
    layers.Flatten(),
    layers.Dense(300, activation="sigmoid"),
    layers.Dense(100, activation="sigmoid"),
    layers.Dense(10, activation="softmax")
])

model_original_relu = models.Sequential([
    layers.Input(shape=(28, 28)),
    layers.Flatten(),
    layers.Dense(300, activation="relu"),
    layers.Dense(100, activation="relu"),
    layers.Dense(10, activation="softmax")
])

model_leaky_relu = keras.models.Sequential([
    keras.layers.Flatten(input_shape=[28, 28]),
    keras.layers.Dense(300),
    keras.layers.LeakyReLU(alpha=0.1),
    keras.layers.Dense(100),
    keras.layers.LeakyReLU(alpha=0.1),
    keras.layers.Dense(10, activation="softmax")
])

model_elu = keras.models.Sequential([
    keras.layers.Flatten(input_shape=[28, 28]),
    keras.layers.Dense(300, activation="elu"),
    keras.layers.Dense(100, activation="elu"),
    keras.layers.Dense(10, activation="softmax")
])

model_selu = keras.models.Sequential([
    keras.layers.Flatten(input_shape=[28, 28]),
    keras.layers.Dense(300, activation="selu"),
    keras.layers.Dense(100, activation="selu"),
    keras.layers.Dense(10, activation="softmax")
])

save_model_results(model_original, "MLP using Sigmoid", lr)
save_model_results(model_original_relu, "MLP using ReLU", lr)
save_model_results(model_selu, "MLP using SELU", lr)
save_model_results(model_leaky_relu , "Leaky ReLU", lr)
save_model_results(model_elu , "ELU", lr)
````

So far the biggest role in increasing the accuracy is changing the activation function from sigmoid to ReLU because we went from 84 to 88. A factor is sigmoid has a vanishing gradient so as the x values increase, the gradients vanish. This causes the accuracy to stall. ReLU avoids the saturation that sigmoid suffers from so gradients remain strong and allow the accuracy to rise from 84 to 88.

It turns out achieving a 96 percent accuracy score using MLP on the Fashion MNIST data is unrealistic because MLP's:
1. flatten the two-dimensional data to one that causes some data to be lost,
2. are prone to overfitting,
3. are sensitive to hyperparameter tuning
4. may require many paramaters
5. have poor translation invariance.

The limitations of MLP's in images are the reason why CNN's were developed to analyze images.
