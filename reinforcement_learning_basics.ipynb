```python
import numpy as np
import random
```

```python
# ----- Simple Gridworld Environment -----
class GridWorld:
    def __init__(self):
        self.size = 4
        self.start = (0, 0)
        self.goal = (3, 3)
        self.reset()

    def reset(self):
        self.pos = self.start
        return self.pos

    def step(self, action):
        x, y = self.pos

        if action == 0:   # up
            x = max(x - 1, 0)
        elif action == 1: # down
            x = min(x + 1, self.size - 1)
        elif action == 2: # left
            y = max(y - 1, 0)
        elif action == 3: # right
            y = min(y + 1, self.size - 1)

        self.pos = (x, y)

        # Rewards
        if self.pos == self.goal:
            return self.pos, 10, True
        else:
            return self.pos, -1, False
```

```python
env = GridWorld()

q_table = np.zeros((env.size, env.size, 4))  # state = (x,y), actions = 4
alpha = 0.1       # learning rate
gamma = 0.99      # discount factor
epsilon = 0.2     # exploration rate
episodes = 500

# ----- Training -----
for ep in range(episodes):
    state = env.reset()
    done = False

    while not done:
        x, y = state

        # epsilon-greedy policy
        if random.random() < epsilon:
            action = random.randint(0, 3)
        else:
            action = np.argmax(q_table[x, y])

        next_state, reward, done = env.step(action)
        nx, ny = next_state

        # Q update
        best_next = np.max(q_table[nx, ny])
        q_table[x, y, action] += alpha * (reward + gamma * best_next - q_table[x, y, action])

        state = next_state

# ----- Demo Run (after training) -----
state = env.reset()
path = [state]

for _ in range(20):
    x, y = state
    action = np.argmax(q_table[x, y])
    state, _, done = env.step(action)
    path.append(state)
    if done:
        break

print("Learned Path:", path)
```
